# **PySpark Tonkenizer**

```python
from pyspark import SparkContext, SparkConf

```

```python
# Configuração do Spark
conf = SparkConf().setAppName("TokenizationApp")
sc = SparkContext(conf=conf)
```

```python
# Função para tokenizar uma linha de texto em palavras
def tokenize(line):
    words = line.split()  # Pode ser necessário ajustar a lógica de tokenização dependendo dos requisitos específicos
    return words
```

```python
# Carregar o arquivo de texto como um RDD (Resilient Distributed Dataset)
text_rdd = sc.textFile("coloque aqui o path de um arquivo txt qualquer")
```

```python
# Aplicar a função de tokenização usando o map
tokenized_rdd = text_rdd.map(tokenize)
```

```python
# Coletar os resultados (isso trará uma lista de listas de palavras)
result = tokenized_rdd.collect()
```

```python
# Imprimir o resultado
for line in result:
    print(line)
```

```python
# Encerrar o contexto do Spark
sc.stop()
```
